## Splitting and Merging Colour Channels

```python
img_NZ_bgr = cv.imread('New_Zealand_Lake.jpg', 1)
b, g, r = cv.split(img_NZ_bgr)

plt.figure(figsize=[20, 5])
plt.subplot(141); plt.imshow(r, cmap='gray'); plt.title('Red Channel');
plt.subplot(142); plt.imshow(g, cmap='gray'); plt.title('Green Channel');
plt.subplot(143); plt.imshow(b, cmap='gray'); plt.title('Blue Channel');

im_merged = cv.merge((b, g, r))
plt.subplot(144); plt.imshow(im_merged[:,:,::-1]); plt.title('Merged Image');
```

### Converting Between Colour Spaces

```python
plt.imshow(img_NZ_bgr)
```

```python
img_NZ_rbg = cv.cvtColor(img_NZ_bgr, cv.COLOR_BGR2RGB)
plt.imshow(img_NZ_rbg)
```

```python
img_hsv = cv.cvtColor(img_NZ_bgr, cv.COLOR_BGR2HSV)

h, s, v = cv.split(img_hsv)

plt.figure(figsize=[20, 5])
plt.subplot(141); plt.imshow(h, cmap='gray'); plt.title('Hue Channel');
plt.subplot(142); plt.imshow(s, cmap='gray'); plt.title('Saturation Channel');
plt.subplot(143); plt.imshow(v, cmap='gray'); plt.title('Value Channel');   
plt.subplot(144); plt.imshow(img_NZ_rbg); plt.title('Original Image');
```

```python
plt.imshow(img_hsv)
```

### Modifying Colour Channels

```python
h_mod = h + 10

plt.figure(figsize=[20, 5])
plt.subplot(141); plt.imshow(h_mod, cmap='gray'); plt.title('New Hue Channel');
plt.subplot(142); plt.imshow(s, cmap='gray'); plt.title('Saturation Channel');
plt.subplot(143); plt.imshow(v, cmap='gray'); plt.title('Value Channel'); 

img_hsv_mod = cv.merge((h_mod, s, v))
img_rgb_mod = cv.cvtColor(img_hsv_mod, cv.COLOR_HSV2RGB)
plt.subplot(144); plt.imshow(img_rgb_mod); plt.title('Modified Image');

plt.figure(figsize=[20, 10])
plt.subplot(121); plt.imshow(img_NZ_rbg); plt.title('Original Image');
plt.subplot(122); plt.imshow(img_rgb_mod); plt.title('Modified Image');
```

## Saving and Writing Images

```python
cv.imwrite('New_Zealand_Lake_Modified.jpg', cv.cvtColor(img_rgb_mod, cv.COLOR_RGB2BGR))

Image(filename='New_Zealand_Lake_Modified.jpg')
```
