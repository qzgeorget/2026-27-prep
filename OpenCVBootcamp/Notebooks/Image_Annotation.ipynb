# Image Annotation

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

import matplotlib
matplotlib.rcParams['figure.figsize'] = (9.0, 9.0)

from PIL import Image
from IPython.display import Image
```

```python
img_bgr = cv.imread("../Images/Apollo_11_Launch.jpg", 1)

img = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

plt.imshow(img)
```

## Draw a line

```python
img_line = img.copy()

cv.line(img_line, (100, 100), (400, 400), (255, 0, 0), thickness = 5, lineType = cv.LINE_AA)

plt.imshow(img_line)
```

## Draw a circle

```python
img_circle = img.copy()

cv.circle(img_circle, (300, 300), 100, (0, 255, 0), thickness = 5, lineType = cv.LINE_AA)

plt.imshow(img_circle)
```

## Draw a rectangle

```python
img_rectangle = img.copy()

cv.rectangle(img_rectangle, (500, 100), (700, 600), (0, 0, 255), thickness = 5, lineType = cv.LINE_8)

plt.imshow(img_rectangle)
```

## Writing text

```python
img_text = img_rectangle.copy()

text = "Apollo 11 Saturn V Launch, July 16, 1969"

fFace = cv.FONT_HERSHEY_PLAIN
fScale = 2.3
fColor = (255, 255, 0)
fThickness = 1

cv.putText(img_text, text, (200, 700), fFace, fScale, fColor, fThickness, cv.LINE_AA) 
plt.imshow(img_text)
```
