# Basic Image Manipulations

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import Image
%matplotlib inline
```

```python
cb_img = cv.imread("OpenCVNotebooks/checkerboard_18x18.png", 0)

plt.imshow(cb_img, cmap='gray')

print(cb_img)
```

```python
# Accessing specific elements in the image
print(cb_img[0, 0])  # Pixel value at (0, 0)
print(cb_img[0, 6])  # Pixel value at (0, 6)
```

## Modifying Image Pixel Value

```python
cb_img_copy = cb_img.copy()
cb_img_copy[2, 2] = 200 
cb_img_copy[2, 3] = 200 
cb_img_copy[3, 2] = 200 
cb_img_copy[3, 3] = 200 

plt.imshow(cb_img_copy, cmap='gray')
print(cb_img_copy)
```

```python
# Use numpy splicing to do the same
cb_img_copy[8:10, 8:10] = 200
plt.imshow(cb_img_copy, cmap='gray')
print(cb_img_copy)
```

## Cropping Images

```python
NZ_boat_img = cv.imread("OpenCVNotebooks/New_Zealand_Boat.jpg", 1)
NZ_boat_img_rgb = cv.cvtColor(NZ_boat_img, cv.COLOR_BGR2RGB)

plt.imshow(NZ_boat_img_rgb)
```

```python
NZ_boat_img_cropped = NZ_boat_img_rgb[200:400, 300:600]
plt.imshow(NZ_boat_img_cropped)
```

## Resizing Images

```python
# By scaling factors
resized_boat_img_2x = cv.resize(NZ_boat_img_cropped, None, fx = 2, fy = 2)
plt.imshow(resized_boat_img_2x)
```

```python
# By desired dimensions
dw = 100
dh = 200
desired_dimensions = (dw, dh)

resized_boat_img_100x200 = cv.resize(NZ_boat_img_cropped, dsize = desired_dimensions, interpolation = cv.INTER_AREA)
plt.imshow(resized_boat_img_100x200)
```

```python
# By dimensions while maintaining aspect ratio

aspect_ratio = NZ_boat_img_cropped.shape[1] / NZ_boat_img_cropped.shape[0]
new_dh = dw / aspect_ratio
new_dimensions = (dw, int(new_dh))
resized_boat_img_aspect_ratio = cv.resize(NZ_boat_img_cropped, dsize = new_dimensions, interpolation = cv.INTER_AREA)
plt.imshow(resized_boat_img_aspect_ratio)
```

## Saving Modified Images

```python
NZ_boat_img_cropped_save = cv.cvtColor(NZ_boat_img_cropped, cv.COLOR_RGB2BGR)
cv.imwrite("OpenCVNotebooks/New_Zealand_Boat_cropped.jpg", NZ_boat_img_cropped_save)

resized_boat_img_2x_save = cv.cvtColor(resized_boat_img_2x, cv.COLOR_RGB2BGR)
cv.imwrite("OpenCVNotebooks/New_Zealand_Boat_cropped_2x.jpg", resized_boat_img_2x_save)

Image(filename="OpenCVNotebooks/New_Zealand_Boat_cropped.jpg")
```

## Flipping Images

```python
NZ_boat_img_hflip = cv.flip(NZ_boat_img_rgb, 1)
NZ_boat_img_vflip = cv.flip(NZ_boat_img_rgb, 0)
NZ_boat_img_hvflip = cv.flip(NZ_boat_img_rgb, -1)

plt.figure(figsize=(18, 5))
plt.subplot(141);plt.imshow(NZ_boat_img_rgb);plt.title("Original")
plt.subplot(142);plt.imshow(NZ_boat_img_hflip);plt.title("Horizontal Flip")
plt.subplot(143);plt.imshow(NZ_boat_img_vflip);plt.title("Vertical Flip")
plt.subplot(144);plt.imshow(NZ_boat_img_hvflip);plt.title("Horizontal & Vertical Flip")
plt.show()
```
