# Writing Videos

```python
import cv2 as cv
import matplotlib.pyplot as plt
%matplotlib inline

source = "../Videos/race_car.mp4"
# source = 0 for webcam

cap = cv.VideoCapture(source)
```

```python
if (cap.isOpened() == False):
    print("Error opening video stream/file.")
```

## Displaying Frames

```python
ret, frame = cap.read()

plt.imshow(frame[...,::-1])
```

```python
from IPython.display import HTML
HTML("""
<video width=1024 controls>
    <source src="../Videos/race_car.mp4" type="video/mp4">
</video>
""")
```

## Writing Video

```python
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
frame_dimensions = (frame_width, frame_height)

# Create empty video writer objects and specify 4cc codec to specify format
out_avi = cv.VideoWriter('../Videos/race_car_out.avi', cv.VideoWriter_fourcc('M', 'J', 'P', 'G'), 10, frame_dimensions)
out_mp4 = cv.VideoWriter('../Videos/race_car_out.mp4', cv.VideoWriter_fourcc(*'XVID'), 10, frame_dimensions)
```

```python
while (cap.isOpened()):
    ret, frame = cap.read()

    if ret == True:
        out_avi.write(frame)
        out_mp4.write(frame)
    else:
        break

cap.release()
out_avi.release()
out_mp4.release()
```
